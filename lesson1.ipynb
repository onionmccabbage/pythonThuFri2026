```python
# we usually import other libraries at the top
import numpy as np # this gives us arrays
import pandas as pd # this gives us dataframes


# we can write comments like this
a = 3
b = 7.3
a
print(a, b)
b
```

```python
# strings are ordinal collections of characters
hello = 'Welcome to Python and stuff'
# we may use slicing
hello[11:17] # start:stop-before
```

```python
# also...
hello[13:18:2] # start:stop:step=1
```

```python
# Other collections
l = [5,4,3,2,1] # a list is an ordinal mutable collection of any data type 
t = (7,6,5,4,3) # a tuple is an immutable ordinal collection of any data type
print( l[3] )
type(l)
l.append(99)
l.insert(3, 42) # also remove, pop and sort
l
type(t)
```

```python
# structures from np and pd
np.random.randint(5)
```

```python
# dict and set
# these are both non ordinal
d = {'n':'Floella', 'assignment':'Dame', 'age':73}
s = {5,3,7,9,2,8,3,5,3} # set will only persist unique members
s
d['n']
print(type(d), type(s))
```

```python
# loops
r = range(0,32,5)
for i in r:
    print(i)
```

```python
for _ in l: # colon : this indicates the start of a code block
    n = _
    m = _ + 3 # + - / * also // integer divide, % remainder after division, ** raise to power
    p = _/3
    print(_, n, m, p)
# when the indentation stops, that is the end of the current code block
```

```python
for (k, v) in d.items():
    print(k, v)
```

```python
# conditionals
# > < == <= >= != and or
l
for i in l:
    if i < 6:
        print('less than six')
    elif i<50:
        print('less than 50')
    elif i<100:
        print('less than 100')
    else:
        print('unkown')
```

```python
# functions
def checkNumeric(n):
    # a docstring
    '''Check if the incoming value n can be made into a number'''
    if type(n) in (int, float):
        return n
    else:
        return 'oops'
```

```python
checkNumeric(3.4364552)
# checkNumeric('coffee')
```
