### Cleaning and Preparing Data

```python
import numpy as np
import pandas as pd
```

```python
# some incomplete data
d = [ 'A', 'B', np.nan, 'D' ]
y = pd.Series(d)
```

```python
y.isna() # isnll, notnull notna
```

```python
# we often use fillna
y.fillna('Q', inplace=True)
```

```python
y
```

```python
# using DataFrame
df = pd.DataFrame(np.random.randn(10, 3))
df.iloc[0:4,1] = np.nan
df.iloc[0:2,2] = np.nan
df.iloc[0,2] = df.iloc[4,0]
df
```

```python
# using fillna
df.describe()
df.fillna( df.mean() )
```
