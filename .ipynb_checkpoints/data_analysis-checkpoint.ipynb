## Financial Analysese

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# statsmodels
import statsmodels.api as sm # linear regression
import statsmodels.tsa as tsa # time series analysis

# import pilo # we may need to add additional libraries
```

```python
# read in our data
sp500 = pd.read_csv('StockData/sp500.csv', parse_dates=['Date'], index_col=['Date'])
aapl = pd.read_csv('StockData/aapl.csv', parse_dates=['Date'], index_col=['Date'])
```

```python
sp500.info()
```

```python
# some calculated fields
sp500['Returns'] = sp500['Close'].pct_change()
aapl['Returns'] = aapl['Close'].pct_change()
# merge the dataframs into one
merged = sp500.merge( aapl, how='inner', left_index=True, right_index=True, suffixes=("_sp", "_aapl") )
```

```python
merged.isna()
# we may check across the entire structure for NA 
df_na = merged[ merged.isna() ]
df_na
```

```python
# we can deal with the NaN
merged.dropna(inplace=True)
```

```python
# Regression model
# CAPM
capm = sm.OLS(merged['Returns_aapl'], merged['Returns_sp'])
capm
```

```python
result = capm.fit()
r2 = result.rsquared
result.summary()
```
