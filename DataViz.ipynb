### Data Visualization
We may use  Pandas, MatPlotLib, Seaborn or Plotly (and many others)

```python
# we aim to gather all the import statements to the top
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load external data
aapl = pd.read_csv( 'StockData/aapl.csv', parse_dates=['Date'], index_col=['Date'] )
#  NB we get a warning about date parsing. May choose to use date_format
sp500 = pd.read_csv( 'StockData/SP500.csv', parse_dates=['Date'], index_col=['Date'] )
```

```python
sp500
```

### Charts

```python
sp500['Close'].plot()
```

### MatPlotLib.pyplot

```python
# remember it is referred to as plt
plt.ylabel('Index Close Price')
plt.xlabel('dates')
plt.title('sfrgasefadthars')
plt.show()
```

```python
# annotating with MAtplotlib
maxPrice = sp500['Close'].max()
# maxDate = sp500['Date'].max()
maxDate = sp500['Close'].idxmax() # retrieve the maximum index value for this column

# annotations
plt.annotate('High', xy=(maxDate, maxPrice))

sp500['Close'].plot()
plt.title('SP500')
plt.plot(maxDate, maxPrice)
```

### Multiline

```python
sp500['Close'].plot()
plt.plot( aapl.index, aapl['Close'], color='teal')
```

### Plotting several charts

```python
# grab more data
jpm = pd.read_csv('StockData/jpm.csv', parse_dates=['Date'], index_col=['Date'])
jnpr = pd.read_csv('StockData/jnpr.csv', parse_dates=['Date'], index_col=['Date'])
```
