### Python with Numpy and Pandas
* Python is very efficient

```python
import numpy as np
import pandas as pd
```

```python
# one-dimensional arrays
a = np.array( [5, 72, 13, 100] )
b = np.array( [2, 5, 10, 30] )

result = np.multiply(a, b)
result
```

```python
# matrices
matrix_a = np.array([[1,2],[3,4]])
matrix_b = np.array([[.5,.6],[.7,.8]])

result = matrix_a * matrix_b
result.T # T will transpose the matrix
```

```python
# we may flatten the matrix
result.flatten()
```

### Slicing and working with columns
Index, iloc and also loc

```python
# we often need to use existing data as an index
# specify an existing column
# sp500.set_index( 'Date' ) # NB this does not persist (by design)
sp500.set_index( 'Date', inplace=True) # here we do persist the change (we lose the original structure)
```

```python
# sp500
# we know about iloc()
sp500.iloc[300:801] # even though the index is 'Date'
# there is still an underlying ordinal index 0, 1, 2 etc
```

```python
# the ONE exception is loc, which DOES include start:stop 
sp500.loc['20130930':'20131007'] # loc will locate, using the index
```

```python
# We may choose to show just some columns
sp500['High']
sp500[ ['Volume', 'High'] ]
```

```python
# using loc and solumn slices together
sp500.loc['20130930':'20131007'][['Open','Close']].mean()
```

### Sorting, Filtering and Calculating new columns

```python
# filtered data
sp500.sort_values( ['Volume', 'Close'], ascending=True )
```

```python
# NB the sort does not persist unless inplace=True
sp500[ sp500['Close'] > sp500['Open'] ]
```

```python
# log values in a new column
sp500['logClose'] = np.log( sp500['Close'] )
sp500
```

```python
# we may calculate the log of returns
sp500['logClose']
sp500['logReturns'] = sp500['logClose'].diff() # default takes previous row
sp500
```

```python
# Moving Average
sp500['ma_1'] = sp500['Close'].shift(1)
sp500['ma_5'] = sp500['ma_1'].rolling(window=5, min_periods=5).mean()
sp500['ma_20'] = sp500['ma_1'].rolling(window=20, min_periods=20).mean()

sp500.head(25)
```

```python
# resample daily, monthly....
rules = {'Open':'first', 'Close':'last', 'Volume':'sum'}
monthlySP500 = sp500.resample('ME').agg(rules)

monthlySP500
```

### Merge Data

```python
# we will load up two deparate sources of data then merge together
aapl = pd.read_csv('StockData/aapl.csv', parse_dates=True, date_format='%Y-%m-%d')
sp500 = pd.read_csv('StockData/sp500.csv', parse_dates=True, date_format='%m/%d/%y') # this will override earier dataframes
aapl['Ticker'] = 'AAPL'
sp500['Ticker'] = 'SP500'
# replace date strings with actual date values
sp500['Date'] = pd.to_datetime( sp500['Date'], format='%m/%d/%y' )
aapl['Date'] = pd.to_datetime( aapl['Date'], format='%Y-%m-%d' )
aapl.info()
```

```python
# we may concaternate these dataFrames
combined = pd.concat([aapl, sp500])
combined # this gives us a lot of concat data
```

```python
merged = sp500.merge(aapl, left_on=['Date'], right_on=['Date'], suffixes=('_sp500', '_aapl'))
```
